```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

def make_rows(n):
    """Artificial tabular data: 4 numeric features + 1 categorical (encoded as int)."""
    X = pd.DataFrame({
        "feat_a": rng.normal(0, 1, n),
        "feat_b": rng.normal(5, 2, n),
        "feat_c": rng.uniform(0, 10, n),
        "feat_d": rng.exponential(1.0, n),
        "category": rng.integers(0, 3, n),
    })
    # Label depends on a noisy combination of features -> learnable binary target
    logit = 1.5 * X["feat_a"] - 0.8 * (X["feat_b"] - 5) + 0.3 * (X["feat_c"] - 5) + 0.7 * (X["category"] == 2)
    p = 1 / (1 + np.exp(-logit))
    y = (rng.uniform(0, 1, n) < p).astype(int)
    return X, y

X_past, y_past = make_rows(500)   # "logged" rows with known labels (training)
X_now, y_now = make_rows(20)      # new rows to predict (y_now kept only for checking)

print(X_past.shape, y_past.shape, X_now.shape)
print("class balance:", np.bincount(y_past))
X_past.head()
```
